```python
from itertools import product

print("P Q R | Q->P | P->~Q | QvR | KB")
print("----------------------------------")

for P, Q, R in product([True, False], repeat=3):
    s1 = (not Q) or P
    s2 = (not P) or (not Q)
    s3 = Q or R

    KB = s1 and s2 and s3

    print(
        int(P), int(Q), int(R), "|",
        int(s1), "    |",
        int(s2), "     |",
        int(s3), "  |",
        int(KB)
    )

print("\nChecking entailment:")

entails_R = True
entails_R_implies_P = True
entails_Q_implies_R = True

for P, Q, R in product([True, False], repeat=3):
    KB = ((not Q) or P) and ((not P) or (not Q)) and (Q or R)

    if KB and not R:
        entails_R = False

    if KB and not ((not R) or P):
        entails_R_implies_P = False

    if KB and not ((not Q) or R):
        entails_Q_implies_R = False

print("KB entails R:", entails_R)
print("KB entails R -> P:", entails_R_implies_P)
print("KB entails Q -> R:", entails_Q_implies_R)
```
