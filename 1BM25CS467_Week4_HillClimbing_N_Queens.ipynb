```python

import random

N = 4

def cost(board):
    attacks = 0
    for i in range(N):
        for j in range(i + 1, N):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                attacks += 1
    return attacks

def hill_climbing():
    board = [1, 1, 1, 1]

    while True:
        current_cost = cost(board)
        print("Board:", board, "Cost:", current_cost)

        if current_cost == 0:
            print("Solution found!")
            return board

        best_board = board[:]
        best_cost = current_cost

        for col in range(N):
            for row in range(1, N + 1):
                if row != board[col]:
                    neighbor = board[:]
                    neighbor[col] = row
                    neighbor_cost = cost(neighbor)

                    if neighbor_cost < best_cost:
                        best_cost = neighbor_cost
                        best_board = neighbor

        if best_cost >= current_cost:
            print("Local minimum reached. Restarting...")
            board = [random.randint(1, N) for _ in range(N)]
        else:
            board = best_board

hill_climbing()
```
